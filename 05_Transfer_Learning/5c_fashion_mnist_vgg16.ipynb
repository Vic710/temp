# 5.c Transfer Learning with VGG16 on Fashion MNIST

**Section 5: Transfer learning**

Object detection using Transfer Learning of CNN architectures for the Fashion
MNIST dataset using the below steps:
a. Set up the environment
b. Loading and preprocessing data
c. Choose the pre-trained Model (Transfer Learning)

d. Freeze the base layers and add custom layers for fine tuning
e. Object detection training
f. Compute loss
g. Evaluate the Model
h. Visualize the prediction

i. Fine Tune the Model: Fine-tuning involves unfreezing the top layers of the pre
trained model and retraining it on the new dataset to adapt it better to your task.

**Note.** `dataset_path` is the original Windows path from the source page and is left unchanged. Point it at a local Fashion MNIST image folder (one subdirectory per class) before running. This notebook does not read the Fashion MNIST CSV.


```python
# 1. Import necessary packages

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import VGG16
```

```python
# 2. Load and preprocess dataset

dataset_path = r"C:\Users\Prajwal\PICT\TE SEM 7\LP3\LP IV Updated Datasets\Object Detection\fashion mnist"

train_data = keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=(224, 224),
    batch_size=32
)

test_data = keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=(224, 224),
    batch_size=32
)


# Get class names

class_names = train_data.class_names
num_classes = len(class_names)

print("Classes:", class_names)
print("Number of Classes:", num_classes)
```

```python
# 3. Choose pretrained VGG16 model

base_model = VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(224, 224, 3)
)
```

```python
# 4. Freeze base model layers

base_model.trainable = False
```

```python
# 5. Define model architecture

model = keras.Sequential([

    layers.Input(shape=(224, 224, 3)),

    layers.Lambda(
        keras.applications.vgg16.preprocess_input
    ),

    base_model,

    layers.Flatten(),

    layers.Dense(256, activation='relu'),

    layers.Dropout(0.5),

    layers.Dense(num_classes, activation='softmax')

])
```

```python
# 6. Compile model

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# 7. Train model

history = model.fit(
    train_data,
    epochs=5,
    validation_data=test_data
)
```

```python
# 8. Compute loss and evaluate model

loss, accuracy = model.evaluate(test_data)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

```python
# 9. Fine Tune the model

base_model.trainable = True

# Freeze all layers except last 4 layers

for layer in base_model.layers[:-4]:
    layer.trainable = False


# Compile model again

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.00001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)


# Fine tune model

history_fine = model.fit(
    train_data,
    epochs=3,
    validation_data=test_data
)
```

```python
# 10. Evaluate fine tuned model

loss, accuracy = model.evaluate(test_data)

print("Fine Tuned Test Loss:", loss)
print("Fine Tuned Test Accuracy:", accuracy)
```

```python
# 11. Visualize predictions

for images, labels in test_data.take(1):

    predictions = model.predict(images)

    plt.figure(figsize=(10, 10))

    for i in range(9):

        plt.subplot(3, 3, i + 1)

        plt.imshow(images[i].numpy().astype("uint8"))

        predicted_class = class_names[
            np.argmax(predictions[i])
        ]

        actual_class = class_names[labels[i].numpy()]

        plt.title(
            "Predicted: " + predicted_class +
            "\nActual: " + actual_class
        )

        plt.axis('off')

    plt.show()
```

```python
# 12. Plot training loss and accuracy

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')

plt.title('Training Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()


plt.subplot(1, 2, 2)

plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')

plt.title('Training Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

plt.show()
```
